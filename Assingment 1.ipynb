## Learn Probability Density Functions using Roll-Number-Parameterized Non-Linear Model

Upload Libraries and Dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
uploaded = files.upload()


```

Load Dataset

```python
df = pd.read_csv("data.csv", encoding='latin1', low_memory=False)
df.head()
```

Selecting Feature and Preprocessing

```python
x = df["no2"] # Select feature

x = x.dropna() # Remove missing values

x = x[x >= 0] # Remove negative values

x.describe()
```

Original Distribution

```python
plt.figure(figsize=(6,3))
plt.hist(x, bins=100, density=True, alpha=0.6)
plt.title("Distribution of NO₂ Concentration")
plt.xlabel("NO₂")
plt.ylabel("Density")
plt.show()
```

# Step-1: 
Transform each value of x into z using the transformation function given below

```python
r = 102313022
a_r = 0.05 * (r % 7)
b_r = 0.3 * ((r % 5) + 1)
a_r, b_r

```

Apply Transformation

```python
z = x + a_r * np.sin(b_r * x)
```

Distribution After Transformation

```python
plt.figure(figsize=(6,3))
plt.hist(z, bins=100, density=True, alpha=0.6, color="blue")
plt.title("Distribution After Non-Linear Transformation (z)")
plt.xlabel("z")
plt.ylabel("Density")
plt.show()
```

# Step-2:
 Learn parameters of the probability density function

```python
mu = np.mean(z)
variance = np.var(z)

lambda_param = 1 / (2 * variance)
c_param = np.sqrt(lambda_param / np.pi)

mu, lambda_param, c_param
```

Visualization

```python
z_range = np.linspace(z.min(), z.max(), 500)
pdf_estimated = c_param * np.exp(-lambda_param * (z_range - mu)**2)

plt.figure(figsize=(8,4))
plt.hist(z, bins=100, density=True, alpha=0.5, label="Distribution")
plt.plot(z_range, pdf_estimated, 'r', linewidth=2, label="Estimated PDF")
plt.legend()
plt.title("Estimated PDF")
plt.show()
```
